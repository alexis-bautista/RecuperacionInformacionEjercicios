### Paso 1: Calcular la matriz TF-IDF para el corpus `data/01_corpus_turismo_500.txt`

En este caso se uso el archivo `data/El arte de amar.txt`

```python
with open("data/El arte de amar.txt", "r", encoding="utf-8") as documento:
    contenido = documento.read()
```

Primero se creo un indice invertido para conocer el numero de veces que aparece un termino en el documento

```python
# convertir todo en minusculas y separar el texto en una lista de palabras
lista_de_palabras = contenido.lower().split()

nombre_documento = "El arte de amar.txt"
indice_invertido = {}

# construccion indice invertido
for palabra in lista_de_palabras:
    
    if palabra not in indice_invertido:
        # Crear palabra en el diccionario principal
        # Su valor será un sub-diccionario con el nombre del doc y el conteo inicial de 1
        indice_invertido[palabra] = {nombre_documento: 1}
        
    else:
        # Acceder a la palabra, luego a su sub-diccionario del documento actual, y sumamos 1
        indice_invertido[palabra][nombre_documento] += 1

print (indice_invertido)
```

Por ejemplo la pabra nota aparece 2 veces en el documento `El arte de amar.txt`

```python
# Representacion en tabla
import pandas as pd

df = pd.DataFrame(list(indice_invertido.items()), columns=['Palabra', 'Archivos'])
df
```

#### **Calculo de la Frecuencia de Terminos (TF)**

```python
import math 
#diccionario para guardar los valores de TF calculados
tf_documento = {}

# recorrer la clave (palabra) y su valor (el sub-diccionario de documentos)
for palabra, documentos in indice_invertido.items():
    
    # frecuencia  (f) buscando el documento en el sub-diccionario
    f = documentos[nombre_documento] 
    
    # formula TF
    tf = 1 + math.log10(f) 
    
    # Guardar el resultado en nuevo diccionario
    tf_documento[palabra] = tf

print(tf_documento)
```

```python
# Representacion en tabla
df_tf = pd.DataFrame(list(tf_documento.items()), columns=['Palabra', 'TF'])
df_tf
```

Por ejemplo la frecuancia del termino 'nota' es 1.301

#### **Calculo de la frecuencia inversa del documento (IDF)**

```python
idf_documento = {}
N = 1 # en esta ocasion se lo define en uno porque el ejemplo se hace en un solo documento

# Recorremos índice invertido
for palabra, documentos in indice_invertido.items():
    
    # Calculo df: ¿En cuántos documentos (claves) aparece la palabra?
    df = len(documentos)
    
    # Calculo IDF 
    idf = math.log10(N / df)
    
    # Guardar el resultado en el diccionario
    idf_documento[palabra] = idf
    
print (idf_documento)
```

```python
# Representacion en tabla
df_idf = pd.DataFrame(list(idf_documento.items()), columns=['Palabra', 'IDF'])
df_idf
```

Como se hace el ejercicio en un solo documento $(N/df_t)=1$ y $log(1)=0$. Lo que se interpreta como terminos demasiados comunes (la diferencia se ve con mas documentos)

#### **Matriz TF - IDF**

```python
tfidf_documento = {}

#Podemos iterar sobre tf_documento porque tiene exactamente las mismas palabras que idf_documento
for palabra in tf_documento:
    
    # Extraer los valores numericos de ambos diccionarios para esa palabra en especifico
    tf = tf_documento[palabra]
    idf = idf_documento[palabra]
    
    # calculo TF-IDF
    w = tf * idf
    
    # Guardar en el nuevo diccionario
    tfidf_documento[palabra] = w

print(tfidf_documento)
```

```python
# Representacion en tabla
df_tfidf = pd.DataFrame(list(tfidf_documento.items()), columns=['Palabra', 'TF-IDF'])
df_tfidf
```

Como el resultado es 0 se concluye que son terminos demasiados comunes, por el mismo hecho de que solo lo hacemos sobre un documento.

### Paso 2: Construir el corpus `Gutenberg 1000`

El corpus `Gutenberg 1000` es un corpus compuesto por 1000 libros de Gutenberg Project

```python
import os

path = "data"
files = os.listdir(path)
N_archivos = len(files)
print("La carpeta",path, "tiene", N_archivos, "documentos")
```

### Paso 3: Calcular la matriz TF-IDF para el corpus `Gutenberg 1000`

Primero se creo un indice invertido para conocer el numero de veces que aparece un termino en los documentos

```python
# diccionario principal
indice_invertido = {}

# Bucle externo: Recorre cada archivo de la lista
for nombre_documento in files:
    
    ruta_completa = os.path.join(path, nombre_documento)
    
    # Abrir y leer el archivo actual
    with open(ruta_completa, "r", encoding="utf-8") as documento:
        contenido = documento.read()
        
    # convertir a minusculas y separar palabraas
    lista_de_palabras = contenido.lower().split()
    
    # Bucle interno: Recorre las palabras del archivo actual
    for palabra in lista_de_palabras:
        
        # Escenario A: La palabra es completamente nueva en todo nuestro índice
        if palabra not in indice_invertido:
            indice_invertido[palabra] = {nombre_documento: 1}
            
        # Escenario B: La palabra ya existe, pero es la primera vez que aparece en ESTE documento en específico
        elif nombre_documento not in indice_invertido[palabra]:
            indice_invertido[palabra][nombre_documento] = 1
            
        # Escenario C: La palabra existe y ya la habíamos visto en ESTE documento, sumamos 1
        else:
            indice_invertido[palabra][nombre_documento] += 1
```

```python
print(indice_invertido.get("amor"))
```

Lo que se interpreta es que la palabra amor aparece 2 veces en el documento `7 de julio.txt`, etc.

```python
# Representacion en tabla
import pandas as pd

df = pd.DataFrame(list(indice_invertido.items()), columns=['Palabra', 'Archivos'])
df
```

#### **Calculo de la Frecuencia de Terminos (TF)**

```python
import math 
#diccionario para guardar los valores de TF calculados
tf_documento = {}

# Bucle externo: recorre la palabra y su sub-diccionario (que contiene varios documentos)
for palabra, documentos in indice_invertido.items():
    
    #sub-diccionario vacío para guardar los TF de esta palabra en específico
    tf_documento[palabra] = {}
    
    # Bucle interno: recorre cada documento donde apareció esta palabra
    # 'nombre_doc' será el nombre del archivo y 'f' será su conteo bruto
    for nombre_doc, f in documentos.items():
        
        # Formula TF
        tf = 1 + math.log10(f) 
        
        # Resultados en un nuevo diccionario
        tf_documento[palabra][nombre_doc] = tf
```

```python
print(tf_documento.get("amor"))
```

Por ejemplo la palabra amor tiene una frecuencia de 1.30 en el documento `7 de julio.txt`

```python
# Representacion en tabla
df_tf = pd.DataFrame(list(tf_documento.items()), columns=['Palabra', 'TF'])
df_tf
```

#### **Calculo de la frecuencia inversa del documento (IDF)**

```python
import math 
idf_documento = {}

for palabra, documentos in indice_invertido.items():
    
    # Calculo df: En cuantos documentos (claves) aparece la palabra?
    df = len(documentos)
    
    # IDF 
    idf = math.log10(N_archivos / df)
    
    # Guardar el resultado 
    idf_documento[palabra] = idf
```

```python
print("IDF de 'amor':", idf_documento.get("amor"))
```

```python
# Representacion en tabla
df_idf = pd.DataFrame(list(idf_documento.items()), columns=['Palabra', 'IDF'])
df_idf
```

#### **Matriz TF - IDF**

```python
tfidf_documento = {}

# Bucle externo: Recorre cada palabra
for palabra in tf_documento:
    
    # Preparamos el sub-diccionario vacío para esta palabra
    tfidf_documento[palabra] = {}
    
    # Extrae el valor numérico del IDF global para esta palabra
    idf = idf_documento[palabra]
    
    # Bucle interno: Recorre los documentos específicos donde aparece esa palabra
    # Aquí 'tf' sí será el valor numérico
    for nombre_doc, tf in tf_documento[palabra].items():
        
        # Calculo TF-IDF
        w = tf * idf
        
        tfidf_documento[palabra][nombre_doc] = w
```

```python
print("TF-IDF de la palabra 'amor':", tfidf_documento.get("amor"))
```

Por ejemplo, la palabra amor tiene un TF-IDF de 0.108 en el documento `7 de julio.txt`

```python
# Representacion en tabla
df_tfidf = pd.DataFrame(list(tfidf_documento.items()), columns=['Palabra', 'TF-IDF'])
df_tfidf
```

### Paso 4: Programar una función `buscar()` para el corpus `Gutenberg 1000`

```python
def buscar(consulta, top_n=5):
    """
    Busca los documentos más relevantes para una consulta dada.
    top_n: Cantidad de resultados a mostrar (por defecto 5)
    """
    # Limpieza y tokenizacion de la consulta
    palabras_consulta = consulta.lower().split()
    
    # Diccionario temporal para acumular la puntuación de los documentos candidatos
    puntuaciones = {}
    
    # Recorremos cada palabra a buscar
    for palabra in palabras_consulta:
        
        # Verificamos si la palabra existe en nuestra matriz TF-IDF
        if palabra in tfidf_documento:
            
            # Extraemos el sub-diccionario (los libros donde aparece y su peso)
            documentos_con_la_palabra = tfidf_documento[palabra]
            
            # Sumamos el peso TF-IDF al acumulado de cada documento
            for nombre_doc, peso in documentos_con_la_palabra.items():
                if nombre_doc not in puntuaciones:
                    puntuaciones[nombre_doc] = 0.0
                
                # Acumulamos el peso (esto es una versión simplificada del producto punto vectorial)
                puntuaciones[nombre_doc] += peso
                
    # Si ninguna palabra de la consulta estaba en los libros
    if not puntuaciones:
        print(f"No se encontraron resultados para: '{consulta}'")
        return
        
    # Ordenar los resultados de mayor a menor puntuación (Ranking)
    # sorted() ordena un diccionario y lo convierte en una lista de tuplas: [('libro1.txt', 2.5), ...]
    resultados_ordenados = sorted(puntuaciones.items(), key=lambda item: item[1], reverse=True)
    
    # Mostrar en pantalla el Top N de resultados
    print(f"--- Top {top_n} resultados para: '{consulta}' ---")
    for posicion, (documento, score) in enumerate(resultados_ordenados[:top_n], start=1):
        print(f"{posicion}. {documento} | Similitud: {score:.4f}")
```

```python
buscar("guerra")
```
